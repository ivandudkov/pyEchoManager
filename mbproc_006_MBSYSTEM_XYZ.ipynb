```python
%load_ext autoreload
%autoreload 2

import os

from filemanager import fileset
from cli import cli

from pyproj import Proj, CRS, Transformer

crs_wgs84 = CRS.from_epsg(4326)
crs_world_mercator = CRS.from_epsg(3395)
transformer = Transformer.from_crs(crs_wgs84, crs_world_mercator, always_xy=True)
```

```python
xyz_file = r'C:\Data\Virtual Machine\SharedFolder\NBP0703\NBP0703_XYZ.xyz'
xyz_output = r'C:\Data\Virtual Machine\SharedFolder\NBP0703\NBP0703_XYZ-filt.xyz'


lines = []
with open(xyz_output, 'w') as f2:
    with open(xyz_file, 'r') as f1:
        f1_content = f1.read().splitlines()
        
        for line in f1_content:
            line_content = line.split('\t')
            lat = float(line_content[1])
            lon = float(line_content[0])
            depth = float(line_content[2])*-1
            if lat < -62.3 or lat > -62.15:
                pass
            elif lon <-59 or lon > -58.5:
                pass
            else:
                
                #lat = float(line_content[0])
                #lon = float(line_content[1])
                #depth = float(line_content[2])
                
                x, y = transformer.transform(lon, lat)
                f2.write(f'{x:.02f}\t{y:.02f}\t{depth:.02f}\n')
        
        
        
        
```
